### Temperature Class

The `Temperature` class represents temperatures that supports Celsius (C), Fahrenheit (F) and Kelvin (K). The class has two attributes, `value` and `unit`. The input for the `unit` attribute is case-insensitive, meaning that both upper- and lowecase units are accepted. These two attributes are also read-only, meaning that they cannot be modified directly. If the unit is not one of the supported units, or if the temperature is below absolute zero (0 Kelvin), a `ValueError` will be raised. 

The class contains four conversion methods:

- `to_celsius()` converts temperatures in Kelvin and Fahrenheit to Celsius.
- `to_kelvin()` converts temperatures in Celsius and Fahrenheit to Kelvin. 
- `to_fahrenheit()` converts Celsius and Kelvin to Fahrenheit.
- `convert_to()` returns a new `Temperature` instance converted to the specified `target_unit`. The target unit is case-insensitive, and a `ValueError` will be raised if the target unit is invalid.

The first three conversion methods return a `float` and `convert_to()` returns a new `Temperature` instance.

The `Temperature` class contains two string representation methods, `__repr__()`, which returns an unambigous representation of the object, and `__str__()` which returns a more human-readable string, formatted with two decimal places. 

Lastly the class contains six comparisons, ==, !=, <, <=, > and >=. We created a `_comapare_value()` helper method to convert the two temperatures to Celsius, so that temperatures could be compared across different units. A tolerance of 1e-9 is used for floating-point compoarisons. The comparison methods returns `NotImplemented` if a `Temperature` instance is compared to a non-`Temperature` instance.

We created a separate Python script, `test_temperature.py` to test the `Temperature class`, we test:

- Initialization
- Validation
- The different conversions
- Round-trip conversions, where we convert a `Temperature` instance from Celsius to Fahrenheit, to Kelvin and back to Celsius, to check that the original temperature is reproduced.
- Cross-unit comparisons
- Formatting
